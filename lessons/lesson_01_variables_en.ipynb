## 1. A variable is a named place for a value

Think of a variable as a **name tag** that helps Python remember a value.

```python
budget = 1000
```

- `budget`: name
- `1000`: value
- `=`: assignment — store the value on the right under the name on the left

````python
budget = 1000
print("Current budget:", budget)
````

## 2. Values have types

| Python type | Meaning | Game example |
|---|---|---|
| `str` | text | unit name |
| `int` | whole number | budget, HP, damage |
| `float` | number that can include a decimal | seconds between battle steps |
| `bool` | either `True` or `False` | show or hide a note |

````python
unit_name = "Tank"
tank_hp = 10
battle_speed = 0.42
show_history_note = True

print(unit_name, type(unit_name))
print(tank_hp, type(tank_hp))
print(battle_speed, type(battle_speed))
print(show_history_note, type(show_history_note))
````

### Important difference

- `tank_cost = 200` stores a number that Python can use in calculations.
- `tank_cost = "200"` stores text inside quotation marks.

They may look similar to us, but Python treats them differently.

````python
starting_budget = 1000
tank_cost = 400

number_of_tanks = starting_budget // tank_cost
money_left = starting_budget % tank_cost

print("Tanks we can buy:", number_of_tanks)
print("Money left:", money_left)
````

## 3. A variable can change

Values such as HP and budget are updated while a game is running. Python calculates the right side first and then stores the result under the name on the left.

````python
tank_hp = 10
anti_tank_damage = 5

print("Before attack:", tank_hp)
tank_hp = tank_hp - anti_tank_damage
print("After one attack:", tank_hp)
tank_hp = tank_hp - anti_tank_damage
print("After two attacks:", tank_hp)
````

### `=` and `==` are different

- `tank_hp = 10`: store 10 under the name `tank_hp`.
- `tank_hp == 10`: ask whether the current value is 10. The answer is `True` or `False`.

````python
tank_hp = 10
print("Is the tank at full HP?", tank_hp == 10)
tank_hp = 5
print("Is the tank at full HP now?", tank_hp == 10)
````

## 7. Why must the game restart?
