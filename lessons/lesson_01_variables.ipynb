## 1. A variable is a named place for a value

변수(variable)는 값을 기억하기 위해 붙이는 **이름표**라고 생각할 수 있습니다.

```python
budget = 1000
```

- `budget`: name
- `1000`: value
- `=`: 오른쪽 값을 왼쪽 이름에 넣으라는 assignment 기호

````python
budget = 1000
print("Current budget:", budget)
````

### Predict first

아래에서 `1000`을 `1500`으로 바꾸면 무엇이 출력될까요? 실행 전에 먼저 말해봅시다.

````python
budget = 1000  # Change only this value, then run the cell again.
print("New budget:", budget)
````

## 2. Values have types

| Python type | 뜻 | 게임 예시 |
|---|---|---|
| `str` | 글자/문장 | unit name |
| `int` | 정수 | budget, HP, damage |
| `float` | 소수도 가능한 수 | seconds between battle steps |
| `bool` | `True` 또는 `False` | show or hide a note |

````python
unit_name = "Tank"
tank_hp = 10
battle_speed = 0.42
show_history_note = True

print(unit_name, type(unit_name))
print(tank_hp, type(tank_hp))
print(battle_speed, type(battle_speed))
print(show_history_note, type(show_history_note))
````

### Important difference

- `tank_cost = 200`은 계산할 수 있는 숫자입니다.
- `tank_cost = "200"`은 따옴표 안의 글자입니다.

겉으로 비슷해 보여도 Python이 다르게 다룹니다.

````python
starting_budget = 1000
tank_cost = 400

number_of_tanks = starting_budget // tank_cost
money_left = starting_budget % tank_cost

print("Tanks we can buy:", number_of_tanks)
print("Money left:", money_left)
````

### Budget challenge

`tank_cost`를 다음 값으로 하나씩 바꾸고, 실행 전에 탱크 수를 예측합니다.

- 500
- 250
- 200

질문: 탱크의 가격이 절반이 되면 살 수 있는 탱크 수는 항상 정확히 두 배가 될까요?

## 3. A variable can change

게임이 진행되면 HP나 budget 같은 값은 업데이트됩니다. 오른쪽 계산을 먼저 하고, 그 결과를 다시 왼쪽 이름에 저장합니다.

````python
tank_hp = 10
anti_tank_damage = 5

print("Before attack:", tank_hp)
tank_hp = tank_hp - anti_tank_damage
print("After one attack:", tank_hp)
tank_hp = tank_hp - anti_tank_damage
print("After two attacks:", tank_hp)
````

### `=` and `==` are different

- `tank_hp = 10`: 10을 `tank_hp`에 저장합니다.
- `tank_hp == 10`: 현재 값이 10인지 물어봅니다. 결과는 `True` 또는 `False`입니다.

````python
tank_hp = 10
print("Is the tank at full HP?", tank_hp == 10)
tank_hp = 5
print("Is the tank at full HP now?", tank_hp == 10)
````

## 6. Scientist loop: Predict → Change → Observe → Explain → Reset

각 실험에서 아래 네 문장을 먼저 채웁니다.

1. **I will change:** `TANK_COST` from 400 to 200.
2. **I predict:** Tank Breaker에서 탱크를 더 많이 살 수 있다.
3. **I observed:** ______________________________
4. **My explanation:** __________________________

한 번에 변수 하나만 바꾸면 어떤 값이 결과를 만들었는지 알 수 있습니다.

### Experiment A — Cheap tanks

1. 게임을 종료합니다.
2. `student_settings.py`에서 `TANK_COST = 400`을 `TANK_COST = 200`으로 바꿉니다.
3. 저장하고 게임을 다시 실행합니다.
4. Tank Breaker를 열어 카드 가격과 살 수 있는 수를 확인합니다.
5. 실험 뒤 400으로 되돌립니다.

### Experiment B — Strong or long-range?

다음 중 **하나만** 선택합니다.

- `TANK_MAX_HP = 20`: 더 오래 살아남을까?
- `TANK_DAMAGE = 6`: 적을 몇 번 만에 격파할까?
- `TANK_RANGE = 4`: 탱크가 언제부터 공격할까?

같은 배치를 사용하여 기본값과 실험값의 결과를 비교합니다.

### Experiment C — Other types

- `TANK_NAME = "Iron Turtle"`로 바꾸면 `str`이 UI를 바꿉니다.
- `SHOW_HISTORY_NOTE = False`로 바꾸면 `bool`이 표시 여부를 바꿉니다.
- `BATTLE_STEP_SECONDS = 0.18`로 바꾸면 `float`이 화면 속도를 바꿉니다.

주의: seconds는 더 작은 값일수록 다음 tick이 더 빨리 와서 전투가 빨라집니다.

## 7. Why must the game restart?

노트북의 `tank_cost`와 설정 파일의 `TANK_COST`는 이름이 비슷해도 서로 다른 Python 실행 공간에 있습니다. 게임은 시작할 때 설정 파일의 값을 읽으므로, 파일을 저장한 뒤 게임을 다시 시작해야 새 값을 사용합니다.
