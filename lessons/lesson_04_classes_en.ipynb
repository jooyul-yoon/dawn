# Lesson 04 — Classes and Objects

Dawn Tactics units are not one giant piece of code. They are a blueprint and many objects made from it. Today we build both.


## Mission brief

Imagine two units that follow the same rules but have their own hit points. When one Tank takes damage, an Infantry unit must not change. Our mission is to give every object its own remembered information.


## Learning goals

- Recall inputs and return values from functions.
- Explain a class as a blueprint and an object as one thing made from it.

Our routine is **Predict → Run → Explain → Change**. Predict the result before running, explain why it happened, then change one value.


```python
def mission_label(objective):
    return "Mission: " + objective


label = mission_label("Hold the bridge")
assert label == "Mission: Hold the bridge"
print(label)
```

## From a function to a blueprint

A function groups an action. A class is a **blueprint** that groups data and actions that use that data. One cookie cutter can make many cookies: one class can make many objects. Their rules match, while each object can remember different values.


```python
class MissionCard:
    def __init__(self, objective):
        self.objective = objective

    def brief(self):
        return "Objective: " + self.objective
```

## Predict the first object

`MissionCard` is still a blueprint. The next cell makes a new object with `MissionCard("Hold the bridge")`. What will `first_card.objective` print? The dot is dot notation: it looks inside one object for an attribute.


```python
first_card = MissionCard("Hold the bridge")
assert first_card.objective == "Hold the bridge"
print(first_card.objective)
```

## `__init__` and `self`

`__init__` is the method that runs when an object is made. `objective` is input for the new object. `self.objective` is an attribute stored inside that object. `self` means the object receiving this method call, so each object can store a different value.


## State of two objects

The first and second cards come from the same class, but they are different objects. The next code changes the first card's objective only. Will the second card change too? Predict on paper, then run it.


```python
second_card = MissionCard("Protect the town")
original_second_objective = second_card.objective
first_card.objective = "Hold the bridge"

assert first_card.objective == "Hold the bridge"
assert second_card.objective == original_second_objective
print(first_card.objective)
print(second_card.objective)
```

## Methods and dot notation

A function written inside a class is a method. Call it after an object and a dot, like `first_card.brief()`. Inside the method, `self.objective` reads the attribute of the card that received the call. Each object has **independent state**: the same blueprint can remember separate names and hit points.


```python
briefing = first_card.brief()
assert briefing == "Objective: Hold the bridge"
print(briefing)
```

```python
from dawn_tactics import Position, Tank, Team


left_tank = Tank(101, Team.BLUE, Position(9, 4))
right_tank = Tank(102, Team.BLUE, Position(9, 6))

left_tank.take_damage(3)

assert left_tank.hp == left_tank.max_hp - 3
assert right_tank.hp == right_tank.max_hp
print("Left Tank:", left_tank.hp, "Right Tank:", right_tank.hp)
```

## Connect to real Dawn Tactics

`left_tank` and `right_tank` above are real game `Tank` objects. Only the left Tank took damage; the right one stayed the same. Now build a smaller `Unit` blueprint yourself: a class line, `__init__`, three attributes, and a method that reduces hit points.


```python
# Write class Unit from its class line.
# Inputs after self: name, max_hp
# Store name, max_hp, and hp. hp starts equal to max_hp.
# take_damage(amount) subtracts damage and never lets hp go below 0.
```

```python
tank = Unit("Tank", 10)
infantry = Unit("Infantry", 5)

assert tank.name == "Tank"
assert tank.max_hp == 10
assert tank.hp == 10
assert infantry.hp == 5

tank.take_damage(3)
assert tank.hp == 7
assert infantry.hp == 5

tank.take_damage(99)
assert tank.hp == 0
print("Mission complete: two objects kept their own state!")
```
