## 4. Return a value

`total_cost(100, 3)`의 두 argument는 unit cost와 count다. 곱한 결과를 `return`하면 calling code가 그 값을 받아 다음 계산에 쓸 수 있다. 이 결과를 **return value**라고 한다.

실행 전에 return value를 계산해 보고, `cost`에 어떤 값이 저장될지 예측한다.


```python
def total_cost(unit_cost, count):
    return unit_cost * count


cost = total_cost(100, 3)
assert cost == 300
print("Total cost:", cost)
```

```python
def print_total(unit_cost, count):
    print(unit_cost * count)


displayed = print_total(100, 3)
returned = total_cost(100, 3)
assert displayed is None
assert returned == 300
print("print_total returned:", displayed)
print("total_cost returned:", returned)
```

## 5. Local variable

`remaining_budget()` 안의 `remaining`은 **local variable**이다. Function이 자기 임무를 수행하는 동안 사용하는 내부 메모이므로 function 밖에서는 같은 이름을 바로 읽을 수 없다.

먼저 `500 - 300`을 예상한 뒤 function을 실행한다. 다음 safety net은 local name을 밖에서 읽을 때 생기는 `NameError`를 잡아 화면에 보여준다. `try/except` 문법은 오늘의 학습 목표가 아니므로 안전장치로만 사용한다.


```python
def remaining_budget(budget, cost):
    remaining = budget - cost
    return remaining


assert remaining_budget(500, 300) == 200
```

```python
try:
    print(remaining)
except NameError as error:
    print("Safety net caught:", type(error).__name__)
else:
    raise AssertionError("remaining should stay local to the function")
```

| Step | Input | Process | Output |
|---|---|---|---|
| `total_cost` | 250, 3 | ______ | ______ |
| `remaining_budget` | 1000, cost | ______ | ______ |

계산을 한 곳에만 정의했기 때문에 같은 곱셈이나 뺄셈을 다시 쓰지 않아도 된다.


```python
cost = total_cost(250, 3)
money_left = remaining_budget(1000, cost)
assert cost == 750
assert money_left == 250
print("Cost:", cost, "Remaining:", money_left)
```

```python
def can_deploy(budget, cost):
    return budget >= cost


assert can_deploy(500, 400) is True
assert can_deploy(300, 400) is False
```
