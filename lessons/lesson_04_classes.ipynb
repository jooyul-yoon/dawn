# Lesson 04 — Class와 Object

Dawn Tactics의 유닛은 하나의 큰 코드 덩어리가 아니라, **설계도와 그 설계도로 만든 여러 object**입니다. 오늘은 class가 무엇인지 직접 만들며 알아봅니다.


## Mission brief

같은 규칙을 따르지만 체력이 서로 다른 두 유닛을 생각해 봅시다. Tank 하나가 피해를 받아도 Infantry의 체력은 바뀌면 안 됩니다. 이처럼 각 물건이 자기 정보를 기억하도록 만드는 것이 오늘 임무입니다.


## 오늘 할 수 있게 될 것

- function으로 했던 입력과 return을 떠올립니다.
- class를 blueprint(설계도), object를 설계도로 만든 한 물건으로 설명합니다.
- `__init__`, `self`, attribute, method, dot notation을 읽고 사용합니다.
- 두 object가 independent state(서로 독립적인 상태)를 가진다는 것을 확인합니다.

오늘의 순서: **Predict → Run → Explain → Change**. 실행 전에 결과를 예측하고, 실행한 뒤 이유를 말한 뒤, 한 값을 바꾸어 봅니다.


```python
def mission_label(objective):
    return "Mission: " + objective


label = mission_label("Hold the bridge")
assert label == "Mission: Hold the bridge"
print(label)
```

## Function에서 blueprint로

function은 일을 묶어 둡니다. class는 **데이터와 그 데이터를 다루는 일을 함께 묶는 blueprint**입니다. 쿠키 틀 하나로 여러 쿠키를 만들 듯, class 하나로 여러 object를 만들 수 있습니다. object마다 기억하는 값은 달라도 규칙은 같습니다.


```python
class MissionCard:
    def __init__(self, objective):
        self.objective = objective

    def brief(self):
        return "Objective: " + self.objective
```

## 첫 object 예측

`MissionCard`는 아직 설계도입니다. 다음 칸에서 `MissionCard("Hold the bridge")`를 실행하면 새 object가 만들어집니다. `first_card.objective`는 무엇을 출력할까요? 점(`.`)은 object 안의 attribute에 찾아 들어가는 dot notation입니다.


```python
first_card = MissionCard("Hold the bridge")
assert first_card.objective == "Hold the bridge"
print(first_card.objective)
```

## `__init__`와 `self`

object를 만들 때 자동으로 시작하는 method가 `__init__`입니다. `objective`는 새 object가 받을 입력입니다. `self.objective`는 그 object 안에 저장되는 attribute입니다. **self는 지금 이 method call을 받은 object를 가리킨다.** 그래서 object마다 다른 값을 저장할 수 있습니다.


## 두 object의 상태

첫 카드와 둘째 카드는 같은 class에서 나왔지만 서로 다른 object입니다. 다음 코드는 첫 카드의 목표만 다시 넣습니다. 둘째 카드의 목표도 따라 바뀔까요? 먼저 종이에 예측하고 실행해 보세요.


```python
second_card = MissionCard("Protect the town")
original_second_objective = second_card.objective
first_card.objective = "Hold the bridge"

assert first_card.objective == "Hold the bridge"
assert second_card.objective == original_second_objective
print(first_card.objective)
print(second_card.objective)
```

## Method와 dot notation

class 안에 쓴 function을 method라고 합니다. `first_card.brief()`처럼 object 뒤에 점을 찍어 부릅니다. method 안의 `self.objective`는 그때 method를 받은 카드의 attribute를 읽습니다. **각 object는 자기만의 state를 가진다.** 같은 blueprint라도 체력과 이름은 각각 따로 기억됩니다.


```python
briefing = first_card.brief()
assert briefing == "Objective: Hold the bridge"
print(briefing)
```

```python
from dawn_tactics import Position, Tank, Team


left_tank = Tank(101, Team.BLUE, Position(9, 4))
right_tank = Tank(102, Team.BLUE, Position(9, 6))

left_tank.take_damage(3)

assert left_tank.hp == left_tank.max_hp - 3
assert right_tank.hp == right_tank.max_hp
print("Left Tank:", left_tank.hp, "Right Tank:", right_tank.hp)
```

## 실제 Dawn Tactics로 연결

위의 `left_tank`와 `right_tank`는 게임 속 진짜 `Tank` object입니다. 왼쪽만 피해를 받았고 오른쪽은 그대로입니다. 이제 같은 생각으로 더 작은 `Unit` blueprint를 직접 만듭니다. 필요한 순서는 class 줄, `__init__`, 세 attribute, 그리고 체력을 줄이는 method입니다.


```python
# Write class Unit from its class line.
# Inputs after self: name, max_hp
# Store name, max_hp, and hp. hp starts equal to max_hp.
# take_damage(amount) subtracts damage and never lets hp go below 0.
```

```python
tank = Unit("Tank", 10)
infantry = Unit("Infantry", 5)

assert tank.name == "Tank"
assert tank.max_hp == 10
assert tank.hp == 10
assert infantry.hp == 5

tank.take_damage(3)
assert tank.hp == 7
assert infantry.hp == 5

tank.take_damage(99)
assert tank.hp == 0
print("Mission complete: two objects kept their own state!")
```
