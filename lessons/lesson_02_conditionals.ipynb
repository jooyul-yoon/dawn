# Lesson 02 — Conditionals: 유닛이 상황을 판단하게 만들기

오늘의 mission: `if`, `elif`, `else`를 사용해 Machine Gun과 Cavalry가 상황에 맞는 damage를 선택하도록 가르칩니다.

## Mission goals

1. condition의 결과가 `True` 또는 `False`임을 설명한다.
2. Python이 어느 branch를 실행할지 먼저 Predict한다.
3. indentation으로 branch에 속한 code를 구분한다.
4. `student_rules.py`의 실제 rule 하나를 바꾸고 결과를 관찰한 뒤 Restore한다.

## 1. Comparison은 질문이다

`target_armor == "LIGHT"`는 yes/no 질문입니다. 다음 cell을 실행하기 전에 두 결과를 먼저 Predict하고, `LIGHT`를 `HEAVY`로 바꿔 다시 확인하세요.

````python
target_armor = "LIGHT"

print("Is LIGHT armor?", target_armor == "LIGHT")
print("Is HEAVY armor?", target_armor == "HEAVY")
````

## 2. `if` and `else`

Python은 먼저 `if`를 확인합니다. 그 결과가 `False`이면 `else`를 실행합니다. 두 branch 중 정확히 하나만 실행됩니다.

````python
target_armor = "LIGHT"

if target_armor == "LIGHT":
    damage = 4
else:
    damage = 1

print("Machine Gun damage:", damage)
````

## 3. `elif`로 선택 하나 더 추가하기

`elif`는 앞의 condition이 `False`일 때 다음 condition을 확인하라는 뜻입니다. `artillery`, `infantry`, `tank`를 차례로 넣어 어느 branch가 실행되는지 확인하세요.

````python
target_kind = "artillery"

if target_kind == "artillery":
    damage = 5
elif target_kind == "infantry":
    damage = 3
else:
    damage = 1

print("Cavalry damage:", damage)
````

## 4. 실제 game rule 읽기

게임은 `src/dawn_tactics/student_rules.py`를 읽습니다. function의 틀은 이미 제공되므로 오늘은 condition과 return damage 값만 편집합니다.

## 5. 모든 branch Preview하기

제공된 safety check는 게임 실행 전에 모든 target을 검사하고, 0부터 999 사이의 whole-number damage가 아니면 멈춥니다.

## Scientist loop

1. **Predict:** 어떤 branch가 실행될지 먼저 말한다.
2. **Change:** condition 하나 또는 return number 하나만 바꾼다.
3. **Preview:** 다섯 가지 preview를 모두 실행한다.
4. **Observe:** 같은 deployment로 게임을 재시작해 관찰한다.
5. **Explain:** 결과를 실행된 branch와 연결해 설명한다.
6. **Reset:** default rule로 Restore한다.
